## Step 8: Convert Date Columns

The sales and inventory datasets contain date columns stored as text.

For proper time-based analysis, these columns need to be converted into datetime format.

This will allow us to analyze monthly sales trends, inventory periods, product movement, and stock risk over time.

```python
# Convert transaction_date in the sales dataset into datetime format
sales["transaction_date"] = pd.to_datetime(sales["transaction_date"])

# Convert start_date in the inventory dataset into datetime format
inventory["start_date"] = pd.to_datetime(inventory["start_date"])
```

## Step 9: Handle Active Inventory Records

The inventory dataset contains an `end_date` column.

Some records may use `9999-12-31` to represent an active or open inventory record rather than a real end date.

To avoid date conversion errors, we first create a flag to identify active inventory records.

```python
# Create a flag for active inventory records where end_date is marked as 9999-12-31
inventory["is_active_inventory_record"] = np.where(
    inventory["end_date"] == "9999-12-31",
    1,
    0)
```

```python
# Replace 9999-12-31 with missing value before converting end_date
inventory["end_date_clean"] = inventory["end_date"].replace("9999-12-31", np.nan)

# Convert cleaned end_date into datetime format
inventory["end_date_clean"] = pd.to_datetime(inventory["end_date_clean"])
```

```python
# Check updated date column data types
sales["transaction_date"].info()

inventory[["start_date", "end_date", "end_date_clean", "is_active_inventory_record"]].info()
```

## Step 12: Check Sales and Inventory Alignment

Before merging sales and inventory data, we need to confirm whether both datasets share common products and stores.

This is important because the final analysis will compare sales demand with inventory availability.

If products or stores exist in one dataset but not the other, the merge may create missing values or incomplete inventory risk analysis.

For this project, this step validates whether `product_no` and `store` can be used as reliable join keys.

```python
# Check product and store counts in both datasets
print("Unique products in sales:", sales["product_no"].nunique())
print("Unique products in inventory:", inventory["product_no"].nunique())

print("Unique stores in sales:", sales["store"].nunique())
print("Unique stores in inventory:", inventory["store"].nunique())
```

```python
# Compare product coverage between sales and inventory
sales_products = set(sales["product_no"].unique())
inventory_products = set(inventory["product_no"].unique())

products_only_in_sales = sales_products - inventory_products
products_only_in_inventory = inventory_products - sales_products

print("Products only in sales:", len(products_only_in_sales))
print("Products only in inventory:", len(products_only_in_inventory))
```

```python
# Compare store coverage between sales and inventory
sales_stores = set(sales["store"].unique())
inventory_stores = set(inventory["store"].unique())

stores_only_in_sales = sales_stores - inventory_stores
stores_only_in_inventory = inventory_stores - sales_stores

print("Stores only in sales:", len(stores_only_in_sales))
print("Stores only in inventory:", len(stores_only_in_inventory))
```

### Sales and Inventory Alignment Finding

The sales and inventory datasets contain matching product and store coverage.

No products or stores were found only in one dataset.

This means the datasets are well aligned and can be merged later using `product_no` and `store` as key fields.

## Step 15: Flag Inventory Data Issues

Sales negative values are linked to product returns, so we keep those records.

Inventory negative values may affect stock and inventory value analysis.

Instead of deleting these records, we create a simple flag to identify inventory rows with possible data issues.

```python
# Flag inventory records with negative stock quantity or negative stock value
inventory["inventory_data_issue_flag"] = np.where(
    (inventory["qty_on_hand"] < 0) |
    (inventory["stocks_selling_amount"] < 0) |
    (inventory["cost_of_stocks"] < 0),
    1,
    0)

# Count inventory records with possible data issues
inventory["inventory_data_issue_flag"].sum()
```

### Inventory Data Issue Finding

The inventory dataset contains 8,026 rows with negative stock quantity or negative stock value.

These records are not removed at this stage. They are flagged using `inventory_data_issue_flag` so they can be reviewed or filtered later during inventory risk analysis.

## Step 16: Create Monthly Sales Features

We create monthly date features from the sales transaction date.

These columns will help analyze sales trends, product movement, and inventory demand by month.

```python
# Create monthly sales features
sales["year_month"] = sales["transaction_date"].dt.to_period("M").astype(str)
sales["month"] = sales["transaction_date"].dt.month

# Preview new monthly features
sales[["transaction_date", "year_month", "month"]].head()
```

## Step 17: Create Sales Business Metrics

We create sales metrics needed for profitability analysis.

These metrics separate normal sales from returns and calculate gross profit for each transaction.

```python
# Create gross sales amount for non-return transactions
sales["gross_sales_amount"] = np.where(
    sales["is_return"] == 0,
    sales["sales_amount"],
    0)

# Create return amount as a positive value
sales["return_amount"] = np.where(
    sales["is_return"] == 1,
    sales["sales_amount"].abs(),
    0)

# Keep net sales amount including returns
sales["net_sales_amount"] = sales["sales_amount"]

# Create gross profit
sales["gross_profit"] = sales["sales_amount"] - sales["cogs"]
```

```python
# Preview new sales metrics
sales[["is_return",
        "qty_sold",
        "sales_amount",
        "cogs",
        "gross_sales_amount",
        "return_amount",
        "net_sales_amount",
        "gross_profit"]].head()
```

## Step 18: Create Inventory Value Metrics

```python
# Create inventory value metrics
inventory["inventory_value_at_cost"] = inventory["cost_of_stocks"]
inventory["inventory_value_at_retail"] = inventory["stocks_selling_amount"]
```

```python
# Preview inventory value metrics
inventory[["qty_on_hand",
        "cost_of_stocks",
        "stocks_selling_amount",
        "inventory_value_at_cost",
        "inventory_value_at_retail"]].head()
```

## Step 19: Create Current Inventory Dataset

We filter the inventory data to active stock records.

This is necessary because the project focuses on current inventory value, cash tied up in stock, profitability,

overstock risk, and products with high stock but weak sales. 

```python
# Keep only active inventory records
current_inventory = inventory[inventory["is_active_inventory_record"] == 1].copy()

# Check current inventory size
current_inventory.shape
```

```python
# Preview current inventory data
current_inventory.head()
```

## Step 20: Aggregate Sales by Product and Store

We summarize sales at the product-store level.

This is necessary because the inventory data is also analyzed by product and store. 

Creating this sales summary allows us to later compare sales performance with current stock availability

```python
# Aggregate sales at product-store level
sales_summary = sales.groupby(["product_no",
        "store",
        "supplier",
        "product_description",
        "product_division",
        "product_category",
        "product_subcategory",
        "product_segment",
        "sales_channel"],
    as_index=False).agg(
    total_gross_sales=("gross_sales_amount", "sum"),
    total_returns=("return_amount", "sum"),
    total_net_sales=("net_sales_amount", "sum"),
    total_cogs=("cogs", "sum"),
    total_gross_profit=("gross_profit", "sum"),
    total_qty_sold=("qty_sold", "sum"),
    total_transactions=("number_of_transactions", "sum"))
```

```python
# Preview sales summary
sales_summary.head()
```

## Step 21: Aggregate Current Inventory by Product and Store

We summarize current inventory at the product-store level.

This prepares the inventory data to be merged with the sales summary,

so we can compare sales performance with current stock availability and inventory value.

```python
# Aggregate current inventory at product-store level
inventory_summary = current_inventory.groupby(["product_no","store"],as_index=False).agg(
    current_qty_on_hand=("qty_on_hand", "sum"),
    current_inventory_value_at_cost=("inventory_value_at_cost", "sum"),
    current_inventory_value_at_retail=("inventory_value_at_retail", "sum"),
    inventory_data_issue_flag=("inventory_data_issue_flag", "max"))
```

```python
# Preview inventory summary
inventory_summary.head()
```

## Step 22: Merge Sales and Current Inventory

We merge product-store sales performance with current inventory.

This creates one analysis dataset that can be used to identify profitable products, overstock risk, stockout risk, and cash tied up in inventory.

```python
# Merge sales summary with current inventory summary
analysis_df = sales_summary.merge(inventory_summary,
    on=["product_no", "store"],
    how="left")
```

```python
# Preview merged dataset
analysis_df.head()
```

```python
# Check merged dataset size
analysis_df.shape
```

## Step 23: Check Missing Values After Merge

After merging sales with current inventory, we check missing values in the full analysis dataset.

## Step 24: Handle Missing Inventory Values

Some product-store records have sales history but no matching current inventory record.

For inventory fields, missing values mean there is no active stock record available. 

We replace these missing inventory values with 0 so inventory risk metrics can be calculated correctly.

```python
# Replace missing inventory values with 0
analysis_df = analysis_df.fillna({"current_qty_on_hand": 0,
    "current_inventory_value_at_cost": 0,
    "current_inventory_value_at_retail": 0,
    "inventory_data_issue_flag": 0})
```

```python
# Confirm missing values are handled
analysis_df.isnull().sum()
```

## Step 25: Create Profitability Metrics

We calculate gross margin percentage to measure profitability.

This helps compare products and stores based on profit efficiency, not only sales revenue.

```python
# Calculate gross margin percentage
analysis_df["gross_margin_pct"] = np.where(analysis_df["total_net_sales"] > 0,
    analysis_df["total_gross_profit"] / analysis_df["total_net_sales"],0)
```

```python
# Preview profitability metric
analysis_df.head()
```

## Step 26: Create Stock-to-Sales Ratio

We calculate the stock-to-sales ratio to compare current stock with sales demand.

This helps identify products with high stock but weak sales, which may indicate overstock risk or cash tied up in inventory.

```python
# Calculate stock-to-sales ratio
analysis_df["stock_to_sales_ratio"] = np.where(
    analysis_df["total_qty_sold"] > 0,
    analysis_df["current_qty_on_hand"] / analysis_df["total_qty_sold"],
    0)
```

```python
# Preview updated analysis dataset
analysis_df.head()
```

## Step 27: Create Inventory Risk Flags

We create simple inventory risk flags using sales demand and current stock.

These flags help identify fast-moving products, slow-moving products, stockout risk, and overstock risk.

```python
# Calculate sales and stock thresholds for risk flags
high_sales_threshold = analysis_df["total_qty_sold"].quantile(0.75)
low_sales_threshold = analysis_df["total_qty_sold"].quantile(0.25)
high_stock_threshold = analysis_df["current_qty_on_hand"].quantile(0.75)
```

```python
# Fast-moving flag:
# Product-store combinations with high quantity sold.
# These items may need frequent replenishment.
analysis_df["fast_moving_flag"] = np.where(
    analysis_df["total_qty_sold"] >= high_sales_threshold,
    1,0)
```

```python
# Slow-moving flag:
# Product-store combinations with low quantity sold.
# These items may need promotion, markdown, or review.
analysis_df["slow_moving_flag"] = np.where(
    analysis_df["total_qty_sold"] <= low_sales_threshold,
    1,0)
```

```python
# Stockout risk flag:
# Product-store combinations with strong sales but no current stock.
# These items may need restocking.
analysis_df["stockout_risk_flag"] = np.where(
    (analysis_df["total_qty_sold"] >= high_sales_threshold) &
    (analysis_df["current_qty_on_hand"] == 0),
    1,0)
```

```python
# Overstock risk flag:
# Product-store combinations with weak sales but high current stock.
# These items may be tying up inventory cash.
analysis_df["overstock_risk_flag"] = np.where(
    (analysis_df["total_qty_sold"] <= low_sales_threshold) &
    (analysis_df["current_qty_on_hand"] >= high_stock_threshold),
    1,
    0)
```

```python
# Check number of records flagged in each risk category
analysis_df[["fast_moving_flag",
        "slow_moving_flag",
        "stockout_risk_flag",
        "overstock_risk_flag"]].sum()
```
